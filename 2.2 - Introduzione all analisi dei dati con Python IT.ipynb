### Pulizia dei dati con `pandas`

Non tutti i dataset sono puliti e ben organizzati fin dall'inizio. A volte parti dei dati mancano o hanno valori illeggibili o strani. In realtà, anche il nostro dataset ha alcuni valori mancanti! Possiamo usare il metodo `.info()` per verificarlo:

```python
data_italy.info()
```

A quanto pare abbiamo molti dati mancanti! Solo 38 anni su 123 hanno un valore presente per la generazione di elettricità. Dobbiamo fare qualcosa al riguardo.

Ci sono due opzioni comuni per gestire i dati mancanti:
* rimozione delle voci mancanti
* imputazione dei dati, ad es. tramite interpolazione o riempimento con un valore

> **NB** Quando si progettano applicazioni di ML è fondamentale scegliere l'approccio corretto per la gestione dei dati mancanti, poiché un approccio sbagliato potrebbe avere un forte impatto sui risultati dei modelli progettati.

Dato che ci sono molte voci mancanti, è più semplice rimuoverle completamente. Per farlo, esegui questo comando:


```python
data_italy.dropna(inplace=True)
# using parameter "inplace=True" automatically saves the new emptied dataset on top of the old one
```

Ora vediamo cosa resta:

```python
data_italy.info()
```

Il dataset finale è limitato alle voci che non sono vuote in **tutte** le colonne.

Ora il dataset è pronto per essere visualizzato. Ottimo!

Per comodità, salverò le trasformazioni che abbiamo fatto come funzione, la useremo più avanti.

```python
def get_country_df(df, country):
    country_df = df[df['country'] == country][list_of_columns].copy()
    return country_df.dropna()
```

### Raggruppare e ordinare i dati in `pandas`

Prima di passare alla visualizzazione ci sono altre due operazioni (molto utili) che `pandas` può gestire: **raggruppamento** e **ordinamento**.

Per spiegare meglio come funzionano dovremmo tornare al grande data frame dell'inizio della lezione.



```python
data.head()
```

Il **raggruppamento** ci permette di prendere un valore _categorico_ come chiave, e calcolare i parametri aggregati che vogliamo. Un buon esempio di variabile _categorica_ può essere la colonna `'country'`.

Immaginiamo ora di voler trovare il PIL pro capite dei paesi nella tabella (e che non esista già una colonna simile nel data frame... :) e confrontarlo con la generazione di elettricità pro capite. Per farlo dobbiamo trovare alcuni valori aggregati di PIL e popolazione per ogni paese. Per semplicità, prendiamo i valori massimi.

Prima creiamo un filtro per le colonne da mantenere:

```python
columns_to_keep = ['gdp', 'population', 'electricity_generation']
```

```python
data.groupby('country')[columns_to_keep].max()  # We imidiately filter to the desired columns
```

Come possiamo vedere, di nuovo abbiamo alcuni valori mancanti. Rimuoviamoli e salviamo il nuovo data frame come `country_data`. Possiamo concatenare i metodi che già conosciamo in un'unica riga. Stampiamo anche le prime righe del nuovo data frame:

```python
country_data = data.groupby('country')[columns_to_keep].max().dropna()
country_data.head()
```

Sembra buono!

Ora aggiungiamo le colonne con i dati che stiamo cercando. L'operazione è molto semplice dato che abbiamo pulito i dati e tutti gli input necessari sono già nel data frame.  

Aggiungiamo una colonna `'gdp_per_capita'` dividendo le colonne `'gdp'` e `'popluation'` e la colonna `'electricity_per_capita'` dividendo le colonne `'electricity_generation'` e `'population'`. Di nuovo, chiamiamo `country_data.head()` per controllare il nostro output.

```python
country_data['gdp_per_capita'] = country_data['gdp'] / country_data['population']
country_data['electricity_per_capita'] = country_data['electricity_generation'] / country_data['population']

country_data.head()
```

Perfetto!

Come ultimo passo possiamo **ordinare** il data frame per avere i paesi con il PIL pro capite più alto in cima. Possiamo farlo usando il metodo `.sort_values()`

> **NB** Il metodo `.sort_values()` richiede alcuni argomenti per funzionare correttamente. L'argomento `by` specifica la chiave di ordinamento, `ascending` indica l'ordine dell'ordinamento. Inoltre, possiamo usare l'argomento (già noto) `inplace` per salvare il dataframe sovrascrivendo quello vecchio.

```python
country_data.sort_values(by='gdp_per_capita', ascending=False)
```

### Esempi di grafici

Altri semplici grafici che possiamo disegnare usando matplotlib includono:
1. Grafici a dispersione (scatter plot)
2. Grafici a barre
3. Istogrammi
4. Mappe di calore
5. Box plot

Il codice seguente mostrerà alcuni esempi di questi grafici. Non devi ancora capire l'intero codice.

**1. Grafico a dispersione**

```python
# List the countries
european_countries = ['Italy', 'Germany', 'United Kingdom', 'France', 'Spain', 'Poland']

# Loop through the chosen countries, select data and plot
for country in european_countries:
    
    data_country = get_country_df(data, country)
    xs = data_country['gdp']
    ys = data_country['electricity_generation']

    plt.scatter(x = xs, y = ys, label = country)

# Plot tidying
plt.title('Electricity generation vs. GDP of European countries')
plt.ylabel('Electricity Generation [TWh]')
plt.xlabel('GDP [trillion USD]')
plt.grid(which='major', alpha = 0.5)

plt.legend()
plt.show()
```

**3. Grafico a barre** 

```python
# List of countries and predeclare lists for holding values
european_countries = ['Italy', 'Germany', 'United Kingdom', 'France', 'Spain', 'Poland']
wind_share = []
solar_share = []

# Loop through countries and get the data (we filter for country and search for the newest value of solar and wind share)
for country in european_countries:
    wind_value = data[data['country'] == country]['wind_share_elec'].iloc[-1]
    wind_share.append(wind_value)

    solar_value = data[data['country'] == country]['solar_share_elec'].iloc[-1]
    solar_share.append(solar_value)

# Define xs and bar heights (the xs are just numbers from 0 to 5 +- half of the width of a bar)
xs_wind = [i-0.2 for i in range(len(wind_share))]
ys_wind = wind_share

xs_solar = [i+0.2 for i in range(len(solar_share))]
ys_solar = solar_share

# Plot
plt.bar(x=xs_wind, height=ys_wind, width=0.4, color='teal', label='Wind')
plt.bar(x=xs_solar, height=ys_solar, width=0.4, color='orange', label='Solar')

# Tidying up
plt.title('Renewable share in electricity generation')
plt.ylabel('Share [%]')
plt.xlabel('Country')
plt.xticks(range(len(wind_share)), european_countries, rotation=-45)
plt.grid(which='major', alpha = 0.5)

plt.legend()
plt.show()
```

**3. Istogrammi**

```python
# Get data
ys = data.groupby('country')['fossil_share_energy'].last().dropna()

# Plot
plt.hist(ys, bins=20, color='brown')

plt.ylim(0, 20)
plt.xlim(0, 100)

# Tidying up
plt.title('Fossil fuels in energy share histogram')
plt.ylabel('Number of countries')
plt.xlabel('Share [%]')
plt.grid(which='major', alpha = 0.5)
plt.yticks(range(0, 21, 5))

plt.show()
```

**4. Box plot**

```python
# List countries
european_countries = ['Italy', 'Germany', 'United Kingdom', 'France', 'Spain', 'Poland', 'Portugal', 'Netherlands', 
                      'Belgium','Luxembourg', 'Austria', 'Hungary', 'Norway', 'Sweden', 'Finland', 'Denmark', 
                      'Iceland', 'Ireland', 'Czechia', 'Slovakia', 'Slovenia', 'Croatia', 'Greece']


fuels = ['Wind', 'Solar', 'Hydro']
colors = ['teal', 'gold', 'lightblue']

# Predeclaration of vars
wind_share = []
solar_share = []
hydro_share = []

# Getting data (the same operation as in the bar plot, simply more countries this time)
for country in european_countries:

    wind_value = data[data['country'] == country]['wind_share_elec'].iloc[-1]
    wind_share.append(wind_value)

    solar_value = data[data['country'] == country]['solar_share_elec'].iloc[-1]
    solar_share.append(solar_value)

    hydro_value = data[data['country'] == country]['hydro_share_elec'].iloc[-1]
    hydro_share.append(hydro_value)
    
fuel_shares = [wind_share, solar_share, hydro_share]


# Plotting
bplot = plt.boxplot(fuel_shares, patch_artist=True, labels = fuels)
for patch, color in zip(bplot['boxes'], colors):
    patch.set_facecolor(color)

# Tidying up
plt.title('Shares of renewables in european electricity generation')
plt.ylabel('Share [%]')
plt.xlabel('Source')
plt.grid(which='major', alpha = 0.5)

plt.show()
```
