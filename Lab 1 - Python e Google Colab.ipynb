### Classi

Le classi sono una specie di super-funzione, che spesso rendono il codice più pulito e con meno errori.

Di solito **inizializziamo** la classe con alcuni dati o attributi.

Poi possiamo chiamare funzioni specifiche della classe, che chiamiamo **metodi**.

````python
class Conductor():
    def __init__(self, resistance):
        self.resistance = resistance
    def calc_loss(self,current):
        return current*current * self.resistance
````

Prima dobbiamo inizializzare un nuovo oggetto a partire dalla classe. Diciamo che è di tipo `Conductor`.

````python
line1 = Conductor(resistance=0.01)
type(line1)
````

Poi possiamo usare l'oggetto e tutti i suoi metodi `def`.

````python
line1.calc_loss(current=5)
````

````python
line1.calc_loss(current=10)
````

### Documentazione

Una buona documentazione riduce gli errori e il tempo necessario per correggerli, quando capitano.

È *estremamente* utile per il tuo io futuro o per un collega.

Cosa succede se passiamo un input non valido, come una stringa? Un errore inaspettato.

````python
line1.calc_loss(current='5')
````

Quindi documentiamo la classe, con commenti `"""` per più righe.

Definiamo anche i tipi dei dati di input e di output. Anche questo riduce gli errori.

````python
class Conductor():
    """ Conductor class """
    def __init__(self, resistance:float):
        """ Conductor useful calculations

        Args:
            resistance (float): resistance of the line in ohms
        """
        self.resistance = resistance

    def calc_loss(self,current:float)->float:
        """ Calculate the line loss

        Args:
            current (float): current of the line in ohms

        Returns:
            float: power in watts
        """
        return current**2 * self.resistance

line2 = Conductor(resistance=0.01)

line2.calc_loss(current=5)
````

Ora se passo il mouse sopra `calc_loss` vedo che l'input non può essere una stringa.

# Calcolo matriciale e parallelizzazione


Il machine learning e l'AI, nel loro nucleo, comportano molta algebra lineare, specialmente moltiplicazioni e addizioni tra matrici.

Per esempio, una rete neurale a singolo strato con 24 unità e 96 valori di input:
- Ogni passaggio in avanti (forward pass) richiede una moltiplicazione matriciale $24\times96$
- Una CPU a singolo thread eseguirà ciascuna delle $24\times96=2304$ moltiplicazioni in sequenza
- Una GPU invece eseguirà molte delle colonne in parallelo, impiegando una frazione del tempo
- L'addestramento della rete neurale richiede _molti_ passaggi in avanti e il calcolo dei gradienti dell'errore

Il vantaggio della GPU
- I grandi modelli di machine learning possono avere milioni di parametri e GB di dati di addestramento
- In molte applicazioni in tempo reale il modello deve calcolare gli output in millisecondi
- Il tempo di addestramento può essere di giorni o settimane
- Il vantaggio della parallelizzazione su GPU può essere un addestramento 5-20 volte più veloce
- Il costo è rappresentato dalla spesa per l'hardware, dalla complessità e dall'energia

Prima e ora
- Le GPU originali furono sviluppate per l'elaborazione grafica (anch'essa una moltiplicazione matriciale veloce e su larga scala)
- Nel frattempo i "cluster" di CPU erano comuni nei laboratori e nelle università
- Ora il principale produttore, NVIDIA, è un elemento critico per lo sviluppo dell'AI
- Ci sono molti livelli di codice che devono funzionare perfettamente con l'hardware GPU per ottenere buone prestazioni
- Tuttavia le nuove CPU hanno più core per _alcune_ operazioni parallele
